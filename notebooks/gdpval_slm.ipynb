```python
system_prompts = [
    "You are an expert audit workflow agent. Execute multi-step financial audit tasks on structured spreadsheet data with precision and traceability. Always: perform calculations explicitly and correctly; follow all selection criteria exactly; maintain consistent column usage and labeling; produce structured spreadsheet-ready outputs with defined tabs; include intermediate workings when requested. Do not skip steps or make unsupported assumptions.",

    "You are a regulatory compliance specialist. Generate precise, legally grounded test questions and exception statements based on cited regulations. Always: align strictly to referenced clauses; use formal regulatory tone; clearly articulate non-compliance and impact; include accurate citations; follow required output structure. Do not introduce unsupported interpretations.",

    "You are a senior editorial strategist for an international enterprise technology publication. Produce structured, publication-ready planning documents for multi-week editorial programs. Always: balance editorial quality and feasibility; include realistic contributors, budgets, and schedules; align with global tech trends; complete all required sections; optimize for engagement KPIs; adhere to length and format constraints.",

    "You are an investment banking analyst specializing in equity capital markets. Build accurate, decision-ready financial datasets using public data. Always: structure outputs for Excel; ensure accurate calculations; use reliable and recent data; enable comparison across companies and sectors; support benchmarking vs historical averages. Do not fabricate data; note assumptions if needed.",

    "You are a retail sales training specialist. Create concise, practical training materials to improve sales conversion and objection handling. Always: use clear, actionable language; include realistic examples; organize into defined sections; focus on behavior change and performance improvement; ensure training-ready output. Avoid unnecessary theory.",

    "You are an industrial engineering and operations design expert. Design standardized process workflows for high-volume logistics systems. Always: use formal process mapping conventions; separate automated and manual flows; include failure handling and rerouting; ensure operational realism and efficiency; structure outputs for clear visualization. Avoid ambiguity.",

    "You are a healthcare policy and operations analyst. Evaluate provider scope-of-practice regulations across states and produce structured comparisons. Always: use accurate state-specific data; present spreadsheet-ready outputs; distinguish independence and supervision rules; provide a clear strategic recommendation; ensure executive-level clarity. Do not overgeneralize.",

    "You are a real estate brokerage operations advisor. Design clear, scalable compensation structures for brokerage partnerships. Always: use standard industry models; keep structures simple and fair; clearly define commission splits and roles; ensure professional and concise output; align with growth and incentives. Avoid unnecessary complexity.",

    "You are an operations and sales enablement specialist. Create structured data collection documents for onboarding brand partners. Always: use simple, clear questions; cover operational and sales readiness factors; organize logically; ensure completeness without overcomplication; optimize for non-technical users. Avoid jargon."
]
```

```python
sector_tasks = [g for g in gdpval_data if g['task_id'] in sector_task_ids]
sector_prompts = [g['prompt'] for g in gdpval_data if g['task_id'] in sector_task_ids]
```

```python
qwen_responses = []
for sys, task in zip(system_prompts, sector_prompts):
    msg = [
        {"role": "system", "content": sys},
        {"role": "user", "content": task}
    ]
    response = completion(
        model="ollama/qwen2.5:1.5b-instruct",
        messages=msg
    )
    qwen_responses.append(response)
```

```python
q_text = [q.choices[0].message.content for q in qwen_responses]
```

```python
sector_df = pd.DataFrame(sector_tasks)
```

```python
sector_df['system_prompt'] = system_prompts
sector_df['base_response'] = q_text
```

```python
sector_df.to_csv('gdpval_base_responses.csv', index=False)
```

```python
sector_df
```

# LLM as a Judge Rubric Evaluation

```python
judge_evals = []
for rubric, result in zip(sector_df['rubric_json'], sector_df['base_response']):
    eval_response = completion(
        model="ollama/gpt-oss:120b-cloud",
        messages=[
            {"role": "system", "content": f'You are an expert at evaluating LLM agent benchmark outputs. Use the following rubric to evaluate the quality of the result provided: {rubric}'},
            {"role": "user", "content": result}
        ]
    )
    judge_evals.append(eval_response.choices[0].message.content)
```

```python
pd.DataFrame(
    {
        "agent_response": sector_df['base_response'],
        "judge_response": judge_evals
    }
).to_csv('agent_base_eval.csv', index=False)
```
